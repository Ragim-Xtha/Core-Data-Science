```python
def ep_count(a):
    check = False
    ep = ""
    for i in a:
        if i == ")":
            break
        
        if i == "(":
            check = True

        if check  == True:
            ep += i


    return ep
```

```python
df["Num_episode"] = df['Title'].apply(ep_count)
```

```python
df.head()
```

```python
df['Num_episode'] = df['Num_episode'].str.replace('(', '', regex=False)
```

```python
df.head()
```

```python
df['Num_episode'] = df['Num_episode'].str.replace(' eps', '')
```

```python
df["Num_episode"] =  df["Num_episode"].astype(int)
```

```python
def time_extract(txt):
    time = ""
    check = False
    for i in range(len(txt)):
        if txt[i] == ")":
            for j in range(i+1,i+20):
                time += txt[j]

            return time

        
```

```python
df["Time"] = df["Title"].apply(time_extract)
df.head()
```

```python
def channel(txt):
    a = ""
    check = False
    for i in txt:
        if i == ":":
            check = True

        if check == True:
            a += i

        if i == "(":
            check = False
    return a
```

```python
df["Channels"] =  df["Title"].apply(channel)
```
