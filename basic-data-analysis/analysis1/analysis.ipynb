### Create a new column score_level
"High" if total_score ≥ 90 , "Medium" if 70 ≤ total_score < 90, "Low" if total_score < 70


```python
def total_score_grade(i):
  if i>=90:
    return "High"
  elif i>70:
    return "Medium"
  elif i<70:
    return "Low"
  else:
    return "NaN"
```

```python
df["grade_score"] = df["total_score"].apply(total_score_grade)
```

```python
def study_cat(i):
  if i<10:
    return "Light"
  elif i>=10 and i<=20:
    return "Moderate"
  elif i>20:
    return "Heavy"
  else:
    return "NaN"
```

```python
df["weekly_self_study_hours_category"] = df['weekly_self_study_hours'].apply(study_cat)
```
