Underweight: Below 18.5

Healthy Weight: 18.5 to 24.9

Overweight: 25.0 to 29.9

Obesity: 30.0 or higher 

```python
def bmi_cat(bmi):
    if bmi <= 19:
        return "Underweight"
    elif bmi<25:
        return "Healthy"
    elif bmi <30:
        return "Overweight"
    elif bmi >=30:
        return "Obesity"
    else:
        return "NO"
```

```python
df["bmi_category"] = df["bmi"].apply(bmi_cat)
```
